```python
import pandas as pd
import re
import unicodedata

# Read existing roster
df = pd.read_csv("colorado_basketball_2027/Roster.csv")

base_url = "https://www.sports-reference.com/req/202512231/cbb/images/players/"

def get_image_url(player):
    # Normalize player name
    player = unicodedata.normalize("NFKD", str(player))
    player = player.encode("ascii", "ignore").decode("ascii")

    # Convert to URL slug
    player = re.sub(r"[^a-z0-9]+", "-", player.lower()).strip("-")

    return f"{base_url}{player}-1.jpg"

# Create player image DataFrame
player_images = pd.DataFrame({
    "Player": df["Player"],
    "Image_URL": df["Player"].apply(get_image_url)
})

# Save to CSV
player_images.to_csv(
    "colorado_basketball_2027/PlayerImageUrls.csv",
    index=False
)

print(player_images)
```
